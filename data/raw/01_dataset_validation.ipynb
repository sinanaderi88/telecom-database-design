## Dataset Structure and Initial Inspection

```python
import pandas as pd

df = pd.read_csv(
    "/home/sina/bigdata-lab/telecom-database-design/data/raw/Telco-Customer-Churn.csv"
)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

## Data Integrity and Value Validation

```python
print("Duplicate Customer IDs:")
print(df["customerID"].duplicated().sum())

print("\nBlank TotalCharges:")
print(df["TotalCharges"].str.strip().eq("").sum())

print("\nTotalCharges Sample:")
print(df["TotalCharges"].head(10).tolist())

print("\nUnique Values:")
for col in ["gender", "InternetService", "Contract",
            "PaymentMethod", "Churn"]:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

## Investigation of Blank TotalCharges Records

```python
blank_charges = df[df["TotalCharges"].str.strip().eq("")]

print(blank_charges[
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Contract"]
])

print("\nTenure distribution:")
print(blank_charges["tenure"].value_counts())
```

## Service Subscription Distribution Analysis

```python
service_columns = [
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

for col in service_columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

## Contract, Payment Method and Numerical Statistics

```python
print("Contract:")
print(df["Contract"].value_counts())

print("\nPayment Method:")
print(df["PaymentMethod"].value_counts())

print("\nMonthly Charges Statistics:")
print(df["MonthlyCharges"].describe())

print("\nTenure Statistics:")
print(df["tenure"].describe())
```
